````python
#@title Function to calculate molecular weight
def calculate_molecular_weight():

    # Defining a dictionary with keys (atomic symbol) and values (molecular mass)
    atomic_masses = {
        'H': 1.008,
        'C': 12.011,
        'N': 14.007,
        'O': 15.999,
        'F': 18.998,
        'Cl': 35.45,
        'Br': 79.904,
        'I': 126.904,
        'He': 4.0026
    }
    print("This function calculates the molecular weight of molecules made up of the following elements:")
    print(', '.join(atomic_masses.keys()),'\n') #each atom name in atomic_masses is called a key
    total_mass = 0.0

    while (True):
        # Ask the user to input an element
        element = input("Enter an element symbol (or type 'done' to finish): ")

        # Check if the user wants to finish input
        if element.lower() == 'done':
            break

        # Check if the element is valid
        if element in atomic_masses:
            # Ask the user how many atoms of this element are present
            try:
                count = int(input(f"How many atoms of {element}? "))
            except:
                print(f"Error: you must enter an integer number of {element} atoms.")
                return("Please try again, and enter an integer number of each atom next time.\n")

            # Add the mass of this element to the total mass
            total_mass += atomic_masses[element] * count
        else:
            print(f"Element {element} is not recognized. Please try again.")

    # Return the total molecular mass
    print("\nThe mass of your molecule is", total_mass, "amu.\n")
    return total_mass
````

````python
#@title Script template for Helium gas simulations
gas_script_template = [
    "units real",
    "dimension 3",
    "atom_style full",
    "pair_style lj/cut 10.0",
    "boundary p p p  # Apply periodic boundary conditions in all three dimensions",
    "",
    "# Create a cubic simulation box (20 Å × 20 Å × 20 Å)",
    "region box block 0 20 0 20 0 20",
    "create_box 1 box",
    "",
    "# Create atoms without overlap constraint",
    "create_atoms 1 random 100 43210 box",
    "",
    "mass 1 4.0026  # Mass of helium in atomic mass units (amu)",
    "pair_coeff * * 0.065 2.64  # Lennard-Jones epsilon and sigma for helium in real units",
    "",
    "thermo 100",
    "thermo_style custom step temp pe ke etotal press",
    "timestep 0.5  # Real units time step (fs)",
    "",
    "# Add a short minimization step to relieve any initial overlaps or high forces",
    "minimize 1.0e-4 1.0e-6 1000 10000",
    "",
    "fix mynve all nve",
    "fix mylgv all langevin 333 333 100.0 1530917",
    "run 10000",
    "",
    "reset_timestep 0",
    "dump myDump all custom 250 helium_VVOLUME_TEMPERATURE.lammps.xyz id type x y z\n",
    "run 50000"
]
````

````python
#@title Function definition: assign_volume_and_temperature()
def assign_volume_and_temperature(volume, temperature, template=gas_script_template):
  out = template.copy()
  length = round(volume**(1/3), 4)
  for index,line in enumerate(out):
    if ('LENGTH' in line) or ('TEMPERATURE' in line) or ('VOLUME' in line):
      newline = line
      if 'LENGTH' in newline:
        newline = newline.replace('LENGTH', str(length))
      if 'TEMPERATURE' in newline:
        newline = newline.replace('TEMPERATURE', str(temperature))
      if 'VOLUME' in newline:
        newline = newline.replace('VOLUME', str(volume))
      out[index] = newline
  return(out)
````

````python
#@title Function definition: lammpsXYZ_to_pymol_xyz()
def lammpsXYZ_to_pymol_xyz(input_file, output_file):
    with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
        timestep = None
        atom_lines = []

        for line in infile:
            line = line.strip()
            if line.startswith("ITEM: TIMESTEP"):
                timestep = int(next(infile).strip())  # Read the timestep value
            elif line.startswith("ITEM: NUMBER OF ATOMS"):
                num_atoms = int(next(infile).strip())  # Read the number of atoms
            elif line.startswith("ITEM: ATOMS"):
                atom_lines = []  # Reset atom lines for the new frame
            elif line.startswith("ITEM:"):
                # Skip other headers like ITEM: BOX BOUNDS
                continue
            else:
                atom_lines.append(line)  # Collect atom data

                # Once all atom lines for the frame are collected, write to the output file
                if len(atom_lines) == num_atoms:
                    # Write in XYZ format: number of atoms, comment line, and atom positions
                    outfile.write(f"{num_atoms}\n")
                    outfile.write(f"Timestep {timestep}\n")
                    for atom_line in atom_lines:
                        atom_info = atom_line.split()
                        atom_id = int(atom_info[0])
                        x, y, z = atom_info[2], atom_info[3], atom_info[4]
                        outfile.write(f"He {x} {y} {z}\n")  # Use "He" as the atom name
                    atom_lines = []  # Reset for next frame
````

````python
#@title Function definition: convert_lammps_to_dataframe()
def convert_lammps_to_dataframe(input_file, output_file):
    # Initialize variables
    data = []

    # Get number of atoms:
    with open(input_file, 'r') as file:
        line = file.readline()
        while line:
            if "ITEM: NUMBER OF ATOMS" in line:
                # Get the next line which contains the number of atoms
                number_of_atoms = int(file.readline().strip())
                print(f"Number of atoms: {number_of_atoms}")
                break  # Stop looping once the value is obtained
            line = file.readline()  # Read the next line

    # Read the input file
    with open(input_file, 'r') as f:
        timestep = None
        for line in f:
            line = line.strip()
            if line.startswith("ITEM: TIMESTEP"):
                timestep = int(next(f).strip())  # Read the timestep value
            elif line.startswith("ITEM: ATOMS"):
                # Read all atoms' data for this timestep
                atom_data = {"Timestep": timestep}
                for atom_line in f:
                    if atom_line.startswith("ITEM") or not atom_line.strip():
                        break
                    atom_info = atom_line.split()
                    atom_id = int(atom_info[0])  # Atom ID
                    x, y, z = map(float, atom_info[2:5])  # Atom coordinates
                    atom_data[f'atom_{atom_id}_x'] = x
                    atom_data[f'atom_{atom_id}_y'] = y
                    atom_data[f'atom_{atom_id}_z'] = z
                data.append(atom_data)

    # Create a DataFrame and save it to CSV
    df = pd.DataFrame(data)
    columns_ordered = [f'atom_{atom_id}_{DIM}' for atom_id in range(1, number_of_atoms+1) for DIM in ['x','y','z'] ]
    df = df[columns_ordered]
    df.to_csv(output_file, index=False)
    return(df)
````

````python
#@title Function definition: animate_2D_trajectory()
def animate_2D_trajectory(df, volume, filename="atom_positions_animation.mp4"):
    sideLength = volume**(1/3)
    # Set up the figure and axis
    fig, ax = plt.subplots()
    scatter = ax.scatter([], [], color='blue', label='He')
    ax.set_xlim(0, sideLength)
    ax.set_ylim(0, sideLength)
    ax.set_aspect('equal')  # Set aspect ratio to 1
    ax.legend(loc='upper right')

    x_columns = [name for name in df.columns if name.endswith('x')]
    z_columns = [name for name in df.columns if name.endswith('z')]

    # Update function for the animation
    def update(frame):
        # Update atom positions based on the frame index
        x_positions = df[x_columns].iloc[frame, :].values
        z_positions = df[z_columns].iloc[frame, :].values
        scatter.set_offsets(np.c_[x_positions, z_positions])  # Update scatter positions
        return [scatter]

    # Create the animation
    ani = FuncAnimation(fig, update, frames=len(df), interval=500, blit=True)

    # Save the animation as a video file
    ani.save(filename, writer='ffmpeg', fps=10)
    plt.close(fig)  # Close the figure to avoid displaying it in Jupyter notebooks
````

````python
#@title Define function to get pressures from output file
def get_pressures(output_file, min_step=10000):
    steps = []
    pressures = []
    with open(output_file, 'r') as file:
        for line in file:
            # Use regex to match lines that have exactly 6 values (Step, Temp, PotEng, KinEng, TotEng, Press)
            match = re.match(r"\s*(\d+)\s+[0-9.E+-]+\s+[-0-9.E+-]+\s+[0-9.E+-]+\s+[0-9.E+-]+\s+([-0-9.E+-]+)", line)
            if match:
                step = int(match.group(1))
                # Append only if step is greater than or equal to min_step
                if step >= min_step:
                    steps.append(step)
                    pressures.append(float(match.group(2)))

    df = pd.DataFrame({
        'Step': steps,
        'Pressure': pressures
    })

    return df

# Example usage:
pressure_data = get_pressures('V10000_T333_He.out')
pressure_data
````
